```python
from collections import Counter
```

```python
# Find duplicates
records = [
    {"contract_id": "A1"},
    {"contract_id": "A2"},
    {"contract_id": "A1"},
    {"contract_id": "A3"},
    {"contract_id": "A2"}
]

seen = set()
duplicates  = set()

for record in records:
    if record['contract_id'] in seen:
        duplicates.add(record['contract_id'])
    else:
        seen.add(record['contract_id'])

print(duplicates)
```

```python
# Count file types
files = [
    "contract1.xml",
    "contract2.xml",
    "award1.pdf",
    "award2.pdf",
    "readme.txt"
]
types =[]
for file in files:
    file_type = file.split(".")[1]
    types.append(file_type)
    file_type_count = Counter(types)

for type, count in file_type_count.items():
    print(f"{type}: {count}")

```

```python
# Aggregate transactions
transactions = [
    ("A", 100),
    ("B", 200),
    ("A", 50),
    ("B", 25)
]

totals = {}
for trans in transactions:
    code = trans[0]
    amount = trans[1]

    if code in totals:
        totals[code] += amount
    else:
        totals[code] = amount 

for code, amount in totals.items():
    print(f"{code}: {amount}")

```
