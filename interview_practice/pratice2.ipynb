#### 1. Find the runner up/ 2nd lowest from a list

```python
# find the runner up from an array
arr = [2,3,6,6,5]

arr.sort(reverse=True)

for i in range(len(arr)):
	if arr[i] != arr[i+1]:
		runner_up=arr[i+1]
		break
	else:
		continue

print(runner_up)
```

```python
def sort_student(name_scores):
	scores = []
	for student in name_scores:
		scores.append(student[1])

	sorted_scores = sorted(set(scores))
	second_lowest = sorted_scores[1]

	names_second_lowest = []
	for student in name_scores:
		if student[1] == second_lowest:
			names_second_lowest.append(student[0])
		names_second_lowest = sorted(names_second_lowest)
	
	for name in names_second_lowest:
		print(name)
```

```python
def student_average(student_marks,query_name):
	for key, value in student_marks.items():
		if key == query_name:
			avg_score = sum(value)/len(value)
	print(f"Average score of {query_name} is {avg_score:.2f}")
```

```python
def mutate_string(string,position,character):

	list_string = list(string)
	list_string[position] = character

	new_str = ''.join(list_string)
	return new_str

s = input()
i,c = input().split()

s_new = mutate_string(s, int(i), c)

print(s_new)
```

#### Substring

```python
string = 'ABCDCDC'
sub_string = 'CDC'

start = 0
end = len(sub_string)

counter = 0

while end <= len(string):
	if sub_string == string[start:end]:
		counter +=1
	start +=1
	end +=1

print(counter)

```

#### 8. Capitalize

```python
name = input()

new_name = name.split()

first_name=new_name[0].capitalize()
print(first_name)
```

```python
def capitalize(string):
	words = []
	for name in string.split():
		words.append(name.capitalize())
	return words

```

```python
n = str(input())

capitalize(n)
```
