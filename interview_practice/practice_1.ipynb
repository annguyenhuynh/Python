```python
"""
lambda arguments:expression
"""

n = int(input())

for i in range(1,n):
	print((lambda i:i*i)(i))
```

```python
def print_consecutive(n):
	final_str = ''

	for i in range(1, n+1):
		final_str = final_str + str(i)
	
	print(final_str)

print_consecutive(5)
```

```python
x=1
y=1
z=2
n=3

for i in range(x+1):
	for j in range(y+1):
		for k in range(z+1):
			if i + j + k !=n:
				print(i,j,k)
```

```python
# List comprehension
new_list = [[i,j,k] for i in range(x+1) for j in range(y+1) for k in range(z+1) if i+j+k!=n]

print(new_list)
```
