```python
import csv
```

```python
with open('names.csv', 'r', newline="") as f:
    csv_reader = csv.reader(f)

    next(csv_reader) # Used to skip header

    for line in csv_reader:
        print(line)
```

```python
# Copying data to a new csv file

with open('names.csv', 'r', newline="") as f:
    csv_reader = csv.reader(f)

    with open('new_names.csv', 'w') as new_f:
        csv_writer = csv.writer(new_f, delimiter='\t')

        for line in csv_reader:
            csv_writer.writerow(line)
```

```python
# Using DictReader and DictWriter
with open('names.csv', 'r') as csv_file:
    csv_reader_1=csv.DictReader(csv_file)

    for line in csv_reader_1:
        print(line)

```

```python
with open('names.csv', 'r') as csv_file:
    csv_reader_1=csv.DictReader(csv_file)

    with open('new_file.csv', 'w') as new_file:
        fieldnames = ['first_name', 'last_name', 'email']

        csv_writer_1 = csv.DictWriter(new_file, fieldnames=fieldnames)

        for line in csv_reader_1:
            csv_writer_1.writerow(line)
```
